# VexarDrive Fleet Analysis

##  Data Preparation

Load the raw workbook, standardise the basic data types, and save the
tables used by the analysis.

The raw workbook is kept unchanged.hanged.

```python
import pandas as pd
import numpy as np

from pathlib import Path

pd.set_option("display.max_columns", None)
```

```python
RAW_PATH = Path("../data/raw/VEXAR_Fleet_Dataset_CANDIDATE_VERSION.xlsx")
PROCESSED_DIR = Path("../data/processed")

PROCESSED_DIR.mkdir(parents=True, exist_ok=True)
```

```python
drivers = pd.read_excel(
    RAW_PATH,
    sheet_name="Drivers",
    header=2
)

vehicles = pd.read_excel(
    RAW_PATH,
    sheet_name="Vehicles",
    header=2
)

trips = pd.read_excel(
    RAW_PATH,
    sheet_name="Trips",
    header=2
)

telemetry = pd.read_excel(
    RAW_PATH,
    sheet_name="Telemetry",
    header=2
)
```

```python
trips["Trip_Date"] = pd.to_datetime(
    trips["Trip_Date"],
    errors="coerce"
)

telemetry["Timestamp"] = pd.to_datetime(
    telemetry["Timestamp"],
    errors="coerce"
)
```

```python
datasets = {
    "drivers": drivers,
    "vehicles": vehicles,
    "trips": trips,
    "telemetry": telemetry,
}

for name, df in datasets.items():
    print(f"{name}: {df.shape}")
```

```python
for name, df in datasets.items():
    df.to_parquet(
        PROCESSED_DIR / f"{name}.parquet",
        index=False
    )
```

```python
processed = {}

for name in datasets:
    processed[name] = pd.read_parquet(
        PROCESSED_DIR / f"{name}.parquet"
    )

for name, df in processed.items():
    print(f"{name}: {df.shape}")
```
